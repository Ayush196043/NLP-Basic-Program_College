```python
# Program 20: Information Extraction using Regular Expressions

import re
import json

# Sample unstructured document
text = """
Customer Name: Rahul Sharma
Email: rahul.sharma@example.com
Phone: +91 98765 43210
Order ID: ORD-12345
Order Date: 24-10-2026
"""

# Extract email addresses
emails = re.findall(
    r"\b[\w.+-]+@[\w.-]+\.[A-Za-z]{2,}\b",
    text
)

# Extract phone numbers
phones = re.findall(
    r"\+?\d[\d ()-]{7,}\d",
    text
)

# Extract order IDs
order_ids = re.findall(
    r"\bORD-\d+\b",
    text,
    flags=re.IGNORECASE
)

# Extract dates
dates = re.findall(
    r"\b\d{1,2}[-/]\d{1,2}[-/]\d{4}\b",
    text
)

# Extract customer name
name_match = re.search(
    r"Customer Name:\s*(.+)",
    text
)

customer_name = name_match.group(1).strip() if name_match else None

# Store extracted information
information = {
    "customer_name": customer_name,
    "emails": emails,
    "phone_numbers": phones,
    "order_ids": order_ids,
    "dates": dates
}

print("Extracted Information:")
print(json.dumps(information, indent=4))
```
