```python
# Program 21: Information Retrieval and Document Ranking using TF-IDF

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Collection of documents
documents = [
    "Python is a programming language used in data science",
    "Deep learning uses neural networks",
    "Natural language processing works with text",
    "TF-IDF helps search engines rank documents",
    "Machine learning is used for text classification"
]

# Convert documents into TF-IDF vectors
vectorizer = TfidfVectorizer(stop_words="english")
document_vectors = vectorizer.fit_transform(documents)

def search_documents(query, top_k=3):

    # Convert search query into a TF-IDF vector
    query_vector = vectorizer.transform([query])

    # Calculate similarity between query and documents
    scores = cosine_similarity(
        query_vector,
        document_vectors
    ).flatten()

    # Sort documents by similarity score
    ranked_indices = scores.argsort()[::-1]

    print("\nSearch Query:", query)
    print("\n===== RANKED SEARCH RESULTS =====")

    result_count = 0

    for index in ranked_indices:

        if scores[index] <= 0:
            continue

        result_count += 1

        print(f"\nRank: {result_count}")
        print(f"Document: {documents[index]}")
        print(f"Similarity Score: {scores[index]:.3f}")

        if result_count >= top_k:
            break

    if result_count == 0:
        print("No relevant documents found.")

# Take query from user
query = input("Enter your search query: ").strip()

if query:
    search_documents(query)
else:
    print("Please enter a valid search query.")
```
