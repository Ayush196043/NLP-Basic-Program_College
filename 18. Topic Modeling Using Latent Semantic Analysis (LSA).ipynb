```python
# Program 18: Topic Modeling using LSA

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD

documents = [
    "Football players score goals in matches",
    "The football team won the match",
    "Players practice football regularly",
    "Python programmers write software",
    "Software developers build applications",
    "Python helps in software development"
]

# Convert text into TF-IDF features
vectorizer = TfidfVectorizer(stop_words="english")
tfidf_matrix = vectorizer.fit_transform(documents)

# Apply Latent Semantic Analysis using SVD
lsa_model = TruncatedSVD(
    n_components=2,
    random_state=42
)

document_topics = lsa_model.fit_transform(tfidf_matrix)

# Get vocabulary
words = vectorizer.get_feature_names_out()

# Display topics
print("===== LSA TOPICS =====")

for topic_number, component in enumerate(lsa_model.components_, start=1):

    top_indices = component.argsort()[-5:][::-1]
    top_words = [words[index] for index in top_indices]

    print(f"Topic {topic_number}: {', '.join(top_words)}")

# Display document coordinates in latent space
print("\n===== DOCUMENT REPRESENTATIONS =====")

for index, representation in enumerate(document_topics, start=1):
    print(f"Document {index}: {representation.round(3)}")
```
