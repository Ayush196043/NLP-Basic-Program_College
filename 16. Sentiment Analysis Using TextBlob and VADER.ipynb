```python
# Program 16: Sentiment Analysis using TextBlob and VADER

from textblob import TextBlob
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

# Sample reviews
reviews = [
    "I absolutely love this product. It is amazing!",
    "The product is okay, nothing special.",
    "This is the worst product. I hate it!"
]

# Initialize VADER analyzer
analyzer = SentimentIntensityAnalyzer()

def classify_sentiment(score):
    if score >= 0.05:
        return "Positive"
    elif score <= -0.05:
        return "Negative"
    else:
        return "Neutral"

for review in reviews:

    # TextBlob sentiment
    blob = TextBlob(review)
    polarity = blob.sentiment.polarity
    subjectivity = blob.sentiment.subjectivity

    # VADER sentiment
    vader_scores = analyzer.polarity_scores(review)
    compound_score = vader_scores["compound"]

    print("\nReview:", review)

    print("TextBlob Polarity:", round(polarity, 3))
    print("TextBlob Subjectivity:", round(subjectivity, 3))

    print("VADER Scores:", vader_scores)
    print("VADER Sentiment:", classify_sentiment(compound_score))
```
