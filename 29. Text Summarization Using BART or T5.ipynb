```python

# Install:
# pip install transformers torch sentencepiece

from transformers import pipeline

summarizer = pipeline(
    "summarization",
    model="facebook/bart-large-cnn"
)

text = """
Artificial intelligence is a branch of computer science.
It focuses on building intelligent machines.
Machine learning is a part of artificial intelligence.
Deep learning uses neural networks to learn patterns.
Natural language processing helps computers understand
and process human language.
AI is used in healthcare, education, banking and business.
"""

summary = summarizer(
    text,
    max_length=60,
    min_length=20,
    do_sample=False
)

print("Original Text:")
print(text)

print("\nSummary:")
print(summary[0]["summary_text"])
```
