```python
# Program 17: Topic Modeling using LDA

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation

documents = [
    "Football players play matches and score goals",
    "The football team won the match",
    "Players practice football every day",
    "Python programmers write software code",
    "Software developers build applications",
    "Python is useful for software development"
]

# Convert documents into a document-term matrix
vectorizer = CountVectorizer(stop_words="english")
document_term_matrix = vectorizer.fit_transform(documents)

# Create LDA model
lda_model = LatentDirichletAllocation(
    n_components=2,
    random_state=42,
    max_iter=20
)

# Train the model
document_topics = lda_model.fit_transform(document_term_matrix)

# Get vocabulary
words = vectorizer.get_feature_names_out()

# Display topics
print("===== DISCOVERED TOPICS =====")

for topic_number, topic in enumerate(lda_model.components_, start=1):

    top_indices = topic.argsort()[-5:][::-1]
    top_words = [words[index] for index in top_indices]

    print(f"Topic {topic_number}: {', '.join(top_words)}")

# Display topic distribution for each document
print("\n===== DOCUMENT TOPIC DISTRIBUTION =====")

for index, distribution in enumerate(document_topics, start=1):
    print(f"Document {index}: {distribution.round(3)}")
```
