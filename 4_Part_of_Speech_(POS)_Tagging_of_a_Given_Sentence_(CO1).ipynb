# 4. Part-of-Speech (POS) Tagging of a Given Sentence (CO1)

```python
import nltk

# Download required packages (run once)
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
sentance = "Ayush is studying Natural Language Processing "
```

```python
words = nltk.word_tokenize(sentance)
```

```python
words
```

```python
pos_tags = nltk.pos_tag(words)
```

```python
pos_tags
```

```python
print("\nPOS Tags:")
for word, tag in pos_tags:
    print(word, "->", tag)
```
