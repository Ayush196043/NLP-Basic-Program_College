```python
# Program 13: Load GloVe Word Embeddings

import numpy as np
from pathlib import Path

# Download glove.6B.50d.txt and place it in this folder
GLOVE_FILE = Path("glove.6B.50d.txt")

def load_glove_embeddings(file_path, selected_words):
    embeddings = {}

    with open(file_path, "r", encoding="utf-8") as file:
        for line in file:
            values = line.strip().split()

            if len(values) < 2:
                continue

            word = values[0]

            if word in selected_words:
                try:
                    vector = np.asarray(values[1:], dtype=np.float32)
                    embeddings[word] = vector
                except ValueError:
                    continue

            if len(embeddings) == len(selected_words):
                break

    return embeddings


words = {"king", "queen", "computer", "language"}

if not GLOVE_FILE.exists():
    print("GloVe file not found.")
    print("Download the GloVe 50-dimensional vectors from:")
    print("https://nlp.stanford.edu/projects/glove/")
else:
    embeddings = load_glove_embeddings(GLOVE_FILE, words)

    for word in sorted(words):
        if word in embeddings:
            print(f"\nWord: {word}")
            print("Vector dimensions:", len(embeddings[word]))
            print("First 10 values:", embeddings[word][:10])
        else:
            print(f"\nWord '{word}' was not found.")
```
