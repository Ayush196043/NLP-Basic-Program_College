```python
# Program 19: Opinion Mining on Product/Service Reviews

import pandas as pd
from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# Expected CSV columns: review, sentiment
file_path = Path("reviews.csv")

if file_path.exists():
    df = pd.read_csv(file_path)

    required_columns = {"review", "sentiment"}

    if not required_columns.issubset(df.columns):
        raise ValueError("CSV must contain review and sentiment columns.")

    df = df[["review", "sentiment"]].dropna()

else:
    # Sample dataset for demonstration
    df = pd.DataFrame({
        "review": [
            "Excellent product and great quality",
            "I love this product",
            "Very useful and reliable",
            "Terrible product quality",
            "I hate this purchase",
            "Very poor customer service",
            "The product is average",
            "It is okay, nothing special",
            "Average quality for the price"
        ],
        "sentiment": [
            "Positive", "Positive", "Positive",
            "Negative", "Negative", "Negative",
            "Neutral", "Neutral", "Neutral"
        ]
    })

print("Dataset:")
print(df.head())

print("\nSentiment Distribution:")
print(df["sentiment"].value_counts())

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    df["review"].astype(str),
    df["sentiment"].astype(str),
    test_size=0.3,
    random_state=42,
    stratify=df["sentiment"]
)

# Build the machine learning pipeline
model = Pipeline([
    ("tfidf", TfidfVectorizer(stop_words="english")),
    ("classifier", LogisticRegression(max_iter=1000))
])

# Train the model
model.fit(X_train, y_train)

# Evaluate
predictions = model.predict(X_test)

print("\nAccuracy:", round(accuracy_score(y_test, predictions), 3))
print("\nClassification Report:")
print(classification_report(y_test, predictions, zero_division=0))

# Predict new reviews
new_reviews = [
    "The product is fantastic",
    "Very bad quality and disappointing",
    "The product is average"
]

print("\nNew Review Predictions:")
for review, prediction in zip(new_reviews, model.predict(new_reviews)):
    print(f"{review} -> {prediction}")
```
