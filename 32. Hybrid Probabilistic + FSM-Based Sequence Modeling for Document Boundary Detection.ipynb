```python

import re

documents = [
    "This is the first document. It contains useful information.",
    "This is another document. It has different content.",
    "Welcome to the new document. This is its content."
]

# Probabilistic boundary detection
def boundary_probability(sentence):
    keywords = ["welcome", "new document", "start"]
    score = 0

    sentence = sentence.lower()

    for keyword in keywords:
        if keyword in sentence:
            score += 1

    return score / len(keywords)


# FSM states
NORMAL = 0
NEW_DOCUMENT = 1

state = NORMAL

for text in documents:

    sentences = re.split(r'(?<=[.!?])\s+', text)

    print("\nProcessing Document:")

    for sentence in sentences:

        probability = boundary_probability(sentence)

        if probability >= 0.33:
            state = NEW_DOCUMENT
        else:
            state = NORMAL

        if state == NEW_DOCUMENT:
            print("\n--- DOCUMENT BOUNDARY DETECTED ---")

        print(sentence)
```
