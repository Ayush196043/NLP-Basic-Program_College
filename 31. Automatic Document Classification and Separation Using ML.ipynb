```python

import os
import shutil

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB

documents = [
    "Python programming and coding",
    "Machine learning and artificial intelligence",
    "Football cricket and sports",
    "Deep learning neural networks",
    "Cricket is my favorite sport",
    "Java programming language"
]

labels = [
    "Programming",
    "AI",
    "Sports",
    "AI",
    "Sports",
    "Programming"
]

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(documents)

model = MultinomialNB()
model.fit(X, labels)

new_documents = [
    "I love Python coding",
    "AI and neural networks",
    "I enjoy playing cricket"
]

os.makedirs("classified_documents", exist_ok=True)

for i, document in enumerate(new_documents):

    vector = vectorizer.transform([document])

    prediction = model.predict(vector)[0]

    folder = os.path.join(
        "classified_documents",
        prediction
    )

    os.makedirs(folder, exist_ok=True)

    file_path = os.path.join(
        folder,
        f"document_{i+1}.txt"
    )

    with open(file_path, "w", encoding="utf-8") as file:
        file.write(document)

    print("Document:", document)
    print("Category:", prediction)
    print("Saved:", file_path)
    print("----------------")
```
