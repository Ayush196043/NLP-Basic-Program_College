# 5. Parsing and Chunking using RegEx and spaCy (CO1)

# **RegEx**

```python
import nltk
import re

# Download required packages
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')

```

```python
# Given sentence
sentence = "Ayush is studying Natural Language Processing at NIET."

# Tokenization
tokens = nltk.word_tokenize(sentence)
```

```python
# POS Tagging
pos_tags = nltk.pos_tag(tokens)

print("POS Tags:")
print(pos_tags)

# RegEx Grammar for Noun Phrase (NP)
grammar = r"""
    NP: {<DT>?<JJ>*<NN.*>+}
"""

# Create RegEx parser
chunk_parser = nltk.RegexpParser(grammar)

# Perform chunking
tree = chunk_parser.parse(pos_tags)

print("\nChunked Tree:")
print(tree)

# Display the tree
tree.pretty_print()
```

# **SpaCy**

```python
import spacy

# Load English spaCy model
nlp = spacy.load("en_core_web_sm")

# Given sentence
sentence = "Ayush is studying Natural Language Processing at NIET."

# Process the sentence
doc = nlp(sentence)

print("Dependency Parsing:\n")

for token in doc:
    print(
        token.text,
        " | POS:", token.pos_,
        " | Dependency:", token.dep_,
        " | Head:", token.head.text
    )
```
