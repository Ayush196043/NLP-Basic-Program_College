```python
# Program 14: Word Mover's Distance using Gensim

import re
from gensim.models import Word2Vec

corpus = [
    "dog chased the ball",
    "puppy played with a ball",
    "cat climbed the tree",
    "machine learning processes data",
    "deep learning uses neural networks",
    "data science analyzes information"
]

# Tokenize corpus
sentences = [
    re.findall(r"\b\w+\b", sentence.lower())
    for sentence in corpus
]

# Train a small Word2Vec model for demonstration
model = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=3,
    min_count=1,
    workers=1,
    epochs=100,
    seed=42
)

# Two similar documents
text1 = "dog chased ball".split()
text2 = "puppy played ball".split()

# One unrelated document
text3 = "machine learning data".split()

# Calculate WMD
distance1 = model.wv.wmdistance(text1, text2)
distance2 = model.wv.wmdistance(text1, text3)

print("Text 1:", text1)
print("Text 2:", text2)
print("Text 3:", text3)

print("\nWMD between Text 1 and Text 2:", distance1)
print("WMD between Text 1 and Text 3:", distance2)

print("\nSmaller WMD generally indicates greater similarity.")
```
