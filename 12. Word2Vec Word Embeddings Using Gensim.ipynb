```python
# Program 12: Word2Vec using Gensim

import re
from gensim.models import Word2Vec

# Custom text corpus
corpus = [
    "natural language processing is interesting",
    "machine learning helps computers learn",
    "word embeddings represent words as vectors",
    "deep learning uses neural networks",
    "natural language processing uses machine learning"
]

# Tokenize the sentences
sentences = [
    re.findall(r"\b\w+\b", sentence.lower())
    for sentence in corpus
]

# Train Word2Vec model
model = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=3,
    min_count=1,
    workers=1,
    sg=1,
    epochs=100,
    seed=42
)

# Display a word vector
word = "language"

print("Word:", word)
print("Word Vector:")
print(model.wv[word])

# Find similar words
print("\nMost Similar Words:")
for similar_word, score in model.wv.most_similar("language", topn=3):
    print(similar_word, round(score, 3))

# Save the trained model
model.save("word2vec_model.model")
print("\nModel saved successfully.")
```
