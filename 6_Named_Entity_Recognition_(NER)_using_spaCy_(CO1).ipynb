# 6. Named Entity Recognition (NER) using spaCy (CO1)


```python
import spacy

# Load English pretrained model
nlp = spacy.load("en_core_web_sm")

# Given text
text = """
Ayush Pandey is studying at Noida Institute of Engineering and Technology
in Greater Noida. He is interested in Natural Language Processing and
Python. He visited Delhi on 10 September 2026.
"""

# Process text
doc = nlp(text)

print("Named Entities:\n")

# Display entities
for ent in doc.ents:
    print(ent.text, " --> ", ent.label_)
```
