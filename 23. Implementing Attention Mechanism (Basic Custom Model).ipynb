```python

import tensorflow as tf
from tensorflow.keras.layers import Layer, Input, Dense, Softmax
from tensorflow.keras.models import Model

class Attention(Layer):

    def __init__(self):
        super().__init__()
        self.score_layer = Dense(1)

    def call(self, inputs):
        scores = self.score_layer(inputs)
        weights = Softmax(axis=1)(scores)

        context = tf.reduce_sum(
            inputs * weights,
            axis=1
        )

        return context

# Input sequence
input_layer = Input(shape=(5, 8))

# Attention
attention_output = Attention()(input_layer)

# Output layer
output = Dense(1, activation="sigmoid")(attention_output)

model = Model(input_layer, output)

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# Sample data
import numpy as np

X = np.random.rand(100, 5, 8)
y = np.random.randint(0, 2, size=(100, 1))

model.fit(X, y, epochs=5, batch_size=8)
```
