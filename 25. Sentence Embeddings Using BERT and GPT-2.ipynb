```python

# Install:
# pip install transformers torch

import torch
from transformers import (
    AutoTokenizer,
    AutoModel,
    GPT2Tokenizer,
    GPT2Model
)

sentences = [
    "Machine learning is interesting",
    "Natural language processing is powerful"
]

# BERT embeddings
bert_tokenizer = AutoTokenizer.from_pretrained(
    "bert-base-uncased"
)

bert_model = AutoModel.from_pretrained(
    "bert-base-uncased"
)

bert_inputs = bert_tokenizer(
    sentences,
    padding=True,
    truncation=True,
    return_tensors="pt"
)

with torch.no_grad():
    bert_output = bert_model(**bert_inputs)

bert_embeddings = bert_output.last_hidden_state.mean(dim=1)

print("BERT Sentence Embeddings:")
print(bert_embeddings.shape)

# GPT-2 embeddings
gpt_tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
gpt_tokenizer.pad_token = gpt_tokenizer.eos_token

gpt_model = GPT2Model.from_pretrained("gpt2")
gpt_model.config.pad_token_id = gpt_tokenizer.pad_token_id

gpt_inputs = gpt_tokenizer(
    sentences,
    padding=True,
    truncation=True,
    return_tensors="pt"
)

with torch.no_grad():
    gpt_output = gpt_model(**gpt_inputs)

mask = gpt_inputs["attention_mask"].unsqueeze(-1)

gpt_embeddings = (
    gpt_output.last_hidden_state * mask
).sum(dim=1) / mask.sum(dim=1)

print("\nGPT-2 Sentence Embeddings:")
print(gpt_embeddings.shape)
```
